```python
import json
from pathlib import Path

mitre_path = Path("../data/raw/mittre/enterprise-attack-19.2.json")

with open(mitre_path, "r", encoding="utf-8") as f:
    mitre_data = json.load(f)

print("Top-level keys:")
print(mitre_data.keys())

print("Number of objects:", len(mitre_data["objects"]))
```

```python
obj = mitre_data["objects"][0]

print(json.dumps(obj, indent=2)[:5000])
```

```python
from collections import Counter

object_types = Counter(
    obj.get("type")
    for obj in mitre_data["objects"]
)

for obj_type, count in object_types.most_common():
    print(f"{obj_type}: {count}")
```

```python
relationships = [
    obj for obj in mitre_data["objects"]
    if obj.get("type") == "relationship"
]

print("Number of relationships:", len(relationships))

print(json.dumps(relationships[0], indent=2)[:3000])
```
